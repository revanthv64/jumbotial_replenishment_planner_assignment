```python
# ============================================================
# CELL 4 — Data type cleaning
# ============================================================

numeric_columns = [
    "vendor_lead_time",
    "current_vendor_mov",
    "max_allocated_space",
    "case_size",
    "cases_allocated",
    "space_value",
    "current_inventory",
    "max_drr",
    "deadweight",
    "orderedquantity",
    "open_po_value",
    "open_po_cases",
    "inv_norm",
    "safety_stock",
    "mrp",
    "cp"
]

for col in numeric_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

# Date
if "earliest_promise_date" in df.columns:
    df["earliest_promise_date"] = pd.to_datetime(
        df["earliest_promise_date"],
        errors="coerce"
    )

print(df.dtypes)
```

```python
# ============================================================
# CELL 5 — Data validation
# ============================================================

print("Number of rows:", len(df))

print("\nDuplicate JPINs:")
print(df["jpin"].duplicated().sum())

print("\nMissing values in important columns:")

important_columns = [
    "jpin",
    "vendor_id",
    "vendor_lead_time",
    "current_inventory",
    "max_drr",
    "inv_norm",
    "safety_stock",
    "max_allocated_space",
    "case_size",
    "cp"
]

missing_summary = (
    df[important_columns]
    .isna()
    .sum()
    .sort_values(ascending=False)
)

display(missing_summary.to_frame("missing_count"))
```

```python
# ============================================================
# CELL 6 — JSON parsing functions
# ============================================================

def safe_json_load(value):
    """
    Safely parse a JSON string.
    Returns {} if the value is blank/invalid.
    """
    if pd.isna(value):
        return {}

    if isinstance(value, dict):
        return value

    try:
        return json.loads(value)
    except (json.JSONDecodeError, TypeError):
        return {}


df["inventory_breakup_parsed"] = df["inventory_breakup"].apply(
    safe_json_load
)

df["open_po_details_parsed"] = df["open_po_details"].apply(
    safe_json_load
)

print("JSON parsing completed.")
```

```python
# ============================================================
# CELL 7 — Extract inventory breakup
# ============================================================

def get_json_number(data, possible_keys):
    """
    Return the first numeric value found among possible keys.
    """
    if not isinstance(data, dict):
        return 0

    # Case-insensitive lookup
    normalized = {
        str(k).lower().replace(" ", "_"): v
        for k, v in data.items()
    }

    for key in possible_keys:
        key_normalized = key.lower().replace(" ", "_")

        if key_normalized in normalized:
            try:
                return float(normalized[key_normalized])
            except (ValueError, TypeError):
                return 0

    return 0


df["sellable_inventory"] = df["inventory_breakup_parsed"].apply(
    lambda x: get_json_number(
        x,
        ["sellable", "sellable_inventory"]
    )
)

df["contingency_inventory"] = df["inventory_breakup_parsed"].apply(
    lambda x: get_json_number(
        x,
        ["contingency"]
    )
)

df["pending_putaway_inventory"] = df["inventory_breakup_parsed"].apply(
    lambda x: get_json_number(
        x,
        ["pending_putaway", "pending_put_away"]
    )
)

df["under_transfer_inventory"] = df["inventory_breakup_parsed"].apply(
    lambda x: get_json_number(
        x,
        ["under_transfer", "transfer"]
    )
)

print(
    df[
        [
            "jpin",
            "current_inventory",
            "sellable_inventory",
            "contingency_inventory",
            "pending_putaway_inventory",
            "under_transfer_inventory"
        ]
    ].head()
)
```

```python
# ============================================================
# CELL 8 — Parse open PO details
# ============================================================

def extract_po_list(data):
    """
    Convert different possible JSON structures into a list of PO records.
    """

    if data is None:
        return []

    if isinstance(data, list):
        return data

    if isinstance(data, dict):

        # Common structures
        for key in ["po_details", "purchase_orders", "orders", "items", "pos"]:
            if key in data and isinstance(data[key], list):
                return data[key]

        # Single PO represented as dictionary
        if any(
            k in data
            for k in [
                "promise_date",
                "promiseDate",
                "quantity",
                "qty"
            ]
        ):
            return [data]

    return []


def get_po_quantity(po):
    """
    Extract quantity from a PO dictionary.
    """

    possible_keys = [
        "quantity",
        "qty",
        "orderedquantity",
        "ordered_quantity",
        "units"
    ]

    for key in possible_keys:
        if key in po:
            try:
                return float(po[key])
            except (ValueError, TypeError):
                return 0

    return 0


def get_po_date(po):
    """
    Extract promise date from a PO dictionary.
    """

    possible_keys = [
        "promise_date",
        "promiseDate",
        "promised_date",
        "delivery_date",
        "expected_date"
    ]

    for key in possible_keys:
        if key in po:
            try:
                return pd.Timestamp(po[key])
            except:
                return pd.NaT

    return pd.NaT
```

```python
# ============================================================
# CELL 9 — Calculate open PO pipeline
# ============================================================

def calculate_open_po_within_lead_time(row):
    """
    Calculate open PO quantity expected to arrive
    on or before the end of vendor lead time.
    """

    pos = extract_po_list(row["open_po_details_parsed"])

    if not pos:
        return 0

    lead_time = row["vendor_lead_time"]

    if pd.isna(lead_time):
        lead_time = 0

    horizon_date = START_DATE + pd.Timedelta(days=float(lead_time))

    total_qty = 0

    for po in pos:

        if not isinstance(po, dict):
            continue

        qty = get_po_quantity(po)
        promise_date = get_po_date(po)

        if pd.isna(promise_date):
            continue

        if promise_date <= horizon_date:
            total_qty += qty

    return total_qty


df["open_po_within_lead_time"] = df.apply(
    calculate_open_po_within_lead_time,
    axis=1
)

print(
    df[
        [
            "jpin",
            "vendor_lead_time",
            "orderedquantity",
            "open_po_within_lead_time"
        ]
    ].head(10)
)
```

```python
# ============================================================
# CELL 10 — Current Days of Inventory
# ============================================================

def calculate_current_doi(row):

    inventory = row["current_inventory"]
    drr = row["max_drr"]

    if pd.isna(inventory):
        inventory = 0

    if pd.isna(drr) or drr <= 0:
        return 0

    return inventory / drr


df["current_days_of_inventory"] = df.apply(
    calculate_current_doi,
    axis=1
)

display(
    df[
        [
            "jpin",
            "current_inventory",
            "max_drr",
            "current_days_of_inventory"
        ]
    ].head()
)
```

```python
# ============================================================
# CELL 11 — Calculate raw replenishment requirement
# ============================================================

def calculate_raw_suggestion(row):

    drr = row["max_drr"]
    lead_time = row["vendor_lead_time"]
    inv_norm = row["inv_norm"]
    safety_stock = row["safety_stock"]

    current_inventory = row["current_inventory"]
    open_po = row["open_po_within_lead_time"]

    # Handle nulls
    drr = 0 if pd.isna(drr) else max(drr, 0)
    lead_time = 0 if pd.isna(lead_time) else max(lead_time, 0)
    inv_norm = 0 if pd.isna(inv_norm) else max(inv_norm, 0)
    safety_stock = 0 if pd.isna(safety_stock) else max(safety_stock, 0)
    current_inventory = (
        0 if pd.isna(current_inventory)
        else max(current_inventory, 0)
    )
    open_po = (
        0 if pd.isna(open_po)
        else max(open_po, 0)
    )

    if drr <= 0:
        return 0

    # Inventory required for lead time
    lead_time_demand = drr * lead_time

    # Inventory norm
    norm_inventory = drr * inv_norm

    # Safety stock
    safety_stock_units = drr * safety_stock

    target_inventory = (
        lead_time_demand
        + norm_inventory
        + safety_stock_units
    )

    available_position = (
        current_inventory
        + open_po
    )

    raw_suggestion = (
        target_inventory
        - available_position
    )

    return max(raw_suggestion, 0)


df["raw_suggestion"] = df.apply(
    calculate_raw_suggestion,
    axis=1
)

display(
    df[
        [
            "jpin",
            "max_drr",
            "vendor_lead_time",
            "inv_norm",
            "safety_stock",
            "current_inventory",
            "open_po_within_lead_time",
            "raw_suggestion"
        ]
    ].head(10)
)
```

```python
# ============================================================
# CELL 12 — Case rounding
# ============================================================

def round_to_cases(row):

    suggestion = row["raw_suggestion"]
    case_size = row["case_size"]

    if pd.isna(suggestion) or suggestion <= 0:
        return 0

    if pd.isna(case_size) or case_size <= 0:
        # Cannot perform case rounding safely
        return 0

    cases = math.ceil(suggestion / case_size)

    return cases


df["initial_cases_suggestion"] = df.apply(
    round_to_cases,
    axis=1
)

df["initial_unit_suggestion"] = (
    df["initial_cases_suggestion"]
    * df["case_size"].fillna(0)
)

display(
    df[
        [
            "jpin",
            "raw_suggestion",
            "case_size",
            "initial_cases_suggestion",
            "initial_unit_suggestion"
        ]
    ].head(10)
)
```

```python
# ============================================================
# CELL 13 — Apply space constraint
# ============================================================

def calculate_final_cases(row):

    required_cases = row["initial_cases_suggestion"]

    if pd.isna(required_cases) or required_cases <= 0:
        return 0

    case_size = row["case_size"]
    max_space = row["max_allocated_space"]

    if pd.isna(case_size) or case_size <= 0:
        return 0

    if pd.isna(max_space) or max_space <= 0:
        return 0

    max_cases_allowed = math.floor(
        max_space / case_size
    )

    return int(
        min(
            required_cases,
            max_cases_allowed
        )
    )


df["final_cases_suggestion"] = df.apply(
    calculate_final_cases,
    axis=1
)

df["final_suggestion"] = (
    df["final_cases_suggestion"]
    * df["case_size"].fillna(0)
)

df["final_suggestion"] = (
    df["final_suggestion"]
    .round()
    .astype(int)
)

df["final_cases_suggestion"] = (
    df["final_cases_suggestion"]
    .astype(int)
)

display(
    df[
        [
            "jpin",
            "initial_unit_suggestion",
            "max_allocated_space",
            "case_size",
            "final_suggestion",
            "final_cases_suggestion"
        ]
    ].head(10)
)
```

```python
# ============================================================
# CELL 14 — Final purchase value
# ============================================================

df["final_value"] = (
    df["final_suggestion"]
    * df["cp"].fillna(0)
)

df["final_value"] = df["final_value"].round(2)

print("Total suggested purchase value:",
      df["final_value"].sum())
```

```python
# ============================================================
# CELL 15 — Final Days of Inventory
# ============================================================

df["projected_inventory"] = (
    df["current_inventory"].fillna(0)
    + df["open_po_within_lead_time"].fillna(0)
    + df["final_suggestion"].fillna(0)
)

def calculate_final_doi(row):

    projected_inventory = row["projected_inventory"]
    drr = row["max_drr"]

    if pd.isna(drr) or drr <= 0:
        return 0

    return projected_inventory / drr


df["final_days_of_inventory"] = df.apply(
    calculate_final_doi,
    axis=1
)

df["final_days_of_inventory"] = (
    df["final_days_of_inventory"]
    .round(2)
)

display(
    df[
        [
            "jpin",
            "current_days_of_inventory",
            "open_po_within_lead_time",
            "final_suggestion",
            "projected_inventory",
            "final_days_of_inventory"
        ]
    ].head(10)
)
```

```python
# ============================================================
# CELL 16 — Final tonnage
# ============================================================

df["final_tonnage"] = (
    df["final_suggestion"].fillna(0)
    * df["deadweight"].fillna(0)
    / 1000
)

df["final_tonnage"] = (
    df["final_tonnage"]
    .round(3)
)

display(
    df[
        [
            "jpin",
            "final_suggestion",
            "deadweight",
            "final_tonnage"
        ]
    ].head()
)
```

```python
# ============================================================
# CELL 17 — MOV validation
# ============================================================

def mov_check(row):

    suggestion = row["final_suggestion"]

    if pd.isna(suggestion) or suggestion <= 0:
        return "N/A"

    criteria = str(
        row["minimum_order_criteria"]
    ).strip().upper()

    mov = row["current_vendor_mov"]

    if pd.isna(mov):
        return "N/A"

    mov = float(mov)

    if criteria == "VALUE":

        return (
            "PASS"
            if row["final_value"] >= mov
            else "FAIL"
        )

    elif criteria == "CASES":

        return (
            "PASS"
            if row["final_cases_suggestion"] >= mov
            else "FAIL"
        )

    return "N/A"


df["mov_check"] = df.apply(
    mov_check,
    axis=1
)

print(df["mov_check"].value_counts(dropna=False))
```

```python
# ============================================================
# CELL 18 — Bonus: Inventory health / priority
# ============================================================

def inventory_priority(row):

    doi = row["current_days_of_inventory"]
    band = str(row["sales_band"]).upper()

    if doi <= 0:
        return "CRITICAL"

    if band == "BAND A":

        if doi < 3:
            return "CRITICAL"
        elif doi < 7:
            return "HIGH"
        elif doi < row["inv_norm"]:
            return "MEDIUM"
        else:
            return "LOW"

    elif band == "BAND B":

        if doi < 3:
            return "HIGH"
        elif doi < row["inv_norm"]:
            return "MEDIUM"
        else:
            return "LOW"

    elif band == "BAND C":

        if doi < 3:
            return "MEDIUM"
        elif doi < row["inv_norm"]:
            return "LOW"
        else:
            return "LOW"

    return "MEDIUM"


df["inventory_priority"] = df.apply(
    inventory_priority,
    axis=1
)

print(df["inventory_priority"].value_counts())
```

```python
# ============================================================
# CELL 19 — Prepare final output
# ============================================================

output_columns = [
    "facility_id",
    "facility_name",
    "jpin",
    "title",
    "category_name",
    "manufacturername",
    "pvname",

    "vendor_id",
    "vendor_name",
    "vendor_lead_time",
    "current_vendor_mov",
    "minimum_order_criteria",

    "inv_norm",
    "safety_stock",

    "current_inventory",
    "max_drr",
    "sales_band",

    "max_allocated_space",
    "case_size",

    "orderedquantity",
    "open_po_value",
    "open_po_cases",
    "earliest_promise_date",

    "mrp",
    "cp",

    "final_suggestion",
    "final_cases_suggestion",
    "final_value",
    "final_days_of_inventory",
    "final_tonnage",
    "mov_check",

    # Bonus / audit columns
    "current_days_of_inventory",
    "open_po_within_lead_time",
    "projected_inventory",
    "inventory_priority"
]

# Keep only columns that exist
output_columns = [
    c for c in output_columns
    if c in df.columns
]

final_df = df[output_columns].copy()

display(final_df.head(10))
```

```python
# ============================================================
# CELL 20 — Create SQLite database
# ============================================================

import sqlite3

# Remove old DB if it exists
db_path = Path(DB_FILE)

if db_path.exists():
    db_path.unlink()

conn = sqlite3.connect(DB_FILE)

print("SQLite database created:", DB_FILE)
```

```python
# ============================================================
# CELL 21 — Load BASE DATA into SQLite
# ============================================================

base_df = pd.read_csv(INPUT_FILE)

# Convert dates to strings for SQLite
if "earliest_promise_date" in base_df.columns:
    base_df["earliest_promise_date"] = pd.to_datetime(
        base_df["earliest_promise_date"],
        errors="coerce"
    ).dt.strftime("%Y-%m-%d")

# Create the table from the base CSV
base_df.to_sql(
    "replenishment_data",
    conn,
    if_exists="replace",
    index=False
)

print("Base data loaded into SQLite.")

result = pd.read_sql(
    "SELECT COUNT(*) AS row_count FROM replenishment_data",
    conn
)

display(result)
```

```python
# ============================================================
# CELL 22 — Add calculated columns
# ============================================================

calculated_columns = {
    "final_suggestion": "INTEGER",
    "final_cases_suggestion": "INTEGER",
    "final_value": "REAL",
    "final_days_of_inventory": "REAL",
    "final_tonnage": "REAL",
    "mov_check": "TEXT",

    # Audit / bonus fields
    "current_days_of_inventory": "REAL",
    "open_po_within_lead_time": "REAL",
    "projected_inventory": "REAL",
    "inventory_priority": "TEXT"
}

existing_columns = pd.read_sql(
    "PRAGMA table_info(replenishment_data)",
    conn
)["name"].tolist()

for column, data_type in calculated_columns.items():

    if column not in existing_columns:

        conn.execute(
            f"""
            ALTER TABLE replenishment_data
            ADD COLUMN {column} {data_type}
            """
        )

conn.commit()

print("Calculated columns added.")
```

```python
# ============================================================
# CELL 23 — Write calculated results back to SQLite
# ============================================================

update_columns = [
    "jpin",
    "final_suggestion",
    "final_cases_suggestion",
    "final_value",
    "final_days_of_inventory",
    "final_tonnage",
    "mov_check",
    "current_days_of_inventory",
    "open_po_within_lead_time",
    "projected_inventory",
    "inventory_priority"
]

update_df = df[update_columns].copy()

# Replace NaN with None so SQLite gets NULL
update_df = update_df.replace({np.nan: None})

update_sql = """
UPDATE replenishment_data
SET
    final_suggestion = ?,
    final_cases_suggestion = ?,
    final_value = ?,
    final_days_of_inventory = ?,
    final_tonnage = ?,
    mov_check = ?,
    current_days_of_inventory = ?,
    open_po_within_lead_time = ?,
    projected_inventory = ?,
    inventory_priority = ?
WHERE jpin = ?
"""

records = []

for _, row in update_df.iterrows():

    records.append(
        (
            row["final_suggestion"],
            row["final_cases_suggestion"],
            row["final_value"],
            row["final_days_of_inventory"],
            row["final_tonnage"],
            row["mov_check"],
            row["current_days_of_inventory"],
            row["open_po_within_lead_time"],
            row["projected_inventory"],
            row["inventory_priority"],
            row["jpin"]
        )
    )

conn.executemany(update_sql, records)
conn.commit()

print("Calculated results written back to SQLite.")
```

```python
# ============================================================
# CELL 24 — Verify database
# ============================================================

verification_query = """
SELECT
    jpin,
    title,
    max_drr,
    current_inventory,
    final_suggestion,
    final_cases_suggestion,
    final_value,
    final_days_of_inventory,
    mov_check
FROM replenishment_data
LIMIT 10
"""

verification_df = pd.read_sql(
    verification_query,
    conn
)

display(verification_df)
```

```python
# ============================================================
# CELL 27 — SQL QUERY 1
# Vendor-level replenishment summary
# ============================================================

query_vendor_summary = """
SELECT
    vendor_id,
    vendor_name,

    COUNT(DISTINCT jpin) AS sku_count,

    SUM(
        CASE
            WHEN final_suggestion > 0
            THEN 1
            ELSE 0
        END
    ) AS skus_requiring_replenishment,

    SUM(final_suggestion) AS total_suggested_units,

    SUM(final_cases_suggestion) AS total_suggested_cases,

    ROUND(
        SUM(final_value),
        2
    ) AS total_suggested_value,

    ROUND(
        SUM(final_tonnage),
        3
    ) AS total_suggested_tonnage,

    SUM(
        CASE
            WHEN mov_check = 'PASS'
            THEN 1
            ELSE 0
        END
    ) AS mov_pass_count,

    SUM(
        CASE
            WHEN mov_check = 'FAIL'
            THEN 1
            ELSE 0
        END
    ) AS mov_fail_count

FROM replenishment_data

GROUP BY
    vendor_id,
    vendor_name

ORDER BY
    total_suggested_value DESC;
"""

vendor_summary = pd.read_sql(
    query_vendor_summary,
    conn
)

display(vendor_summary)
```

```python
# ============================================================
# CELL 28 — SQL QUERY 2
# Top 10 riskiest SKUs
# ============================================================

query_riskiest_skus = """
SELECT
    facility_id,
    facility_name,

    vendor_id,
    vendor_name,

    jpin,
    title,

    max_drr,
    current_inventory,

    ROUND(
        CAST(current_inventory AS REAL)
        / NULLIF(max_drr, 0),
        2
    ) AS current_days_of_inventory,

    inv_norm,
    safety_stock,

    final_suggestion,
    final_days_of_inventory,

    inventory_priority

FROM replenishment_data

WHERE
    max_drr > 0

ORDER BY
    current_days_of_inventory ASC

LIMIT 10;
"""

riskiest_skus = pd.read_sql(
    query_riskiest_skus,
    conn
)

display(riskiest_skus)
```

```python
# ============================================================
# CELL 29 — BONUS SQL QUERY
# Sales-band inventory health
# ============================================================

query_sales_band = """
SELECT
    sales_band,

    COUNT(DISTINCT jpin) AS sku_count,

    SUM(
        CASE
            WHEN final_suggestion > 0
            THEN 1
            ELSE 0
        END
    ) AS replenishment_sku_count,

    SUM(final_suggestion) AS suggested_units,

    SUM(final_cases_suggestion) AS suggested_cases,

    ROUND(
        SUM(final_value),
        2
    ) AS suggested_value,

    ROUND(
        AVG(current_days_of_inventory),
        2
    ) AS avg_current_doi,

    SUM(
        CASE
            WHEN inventory_priority = 'CRITICAL'
            THEN 1
            ELSE 0
        END
    ) AS critical_sku_count

FROM replenishment_data

GROUP BY sales_band

ORDER BY
    suggested_value DESC;
"""

sales_band_summary = pd.read_sql(
    query_sales_band,
    conn
)

display(sales_band_summary)
```

```python
# ============================================================
# CELL 30 — BONUS SQL QUERY
# MOV failures
# ============================================================

query_mov_failures = """
SELECT
    vendor_id,
    vendor_name,
    jpin,
    title,
    minimum_order_criteria,
    current_vendor_mov,
    final_cases_suggestion,
    final_value,
    mov_check

FROM replenishment_data

WHERE mov_check = 'FAIL'

ORDER BY
    final_value DESC;
"""

mov_failures = pd.read_sql(
    query_mov_failures,
    conn
)

display(mov_failures)
```

```python
# ============================================================
# CELL 32 — VALIDATION: Space constraint
# ============================================================

space_violations = df[
    df["final_suggestion"]
    > df["max_allocated_space"].fillna(0)
]

print(
    "Space violations:",
    len(space_violations)
)

if len(space_violations) > 0:
    display(
        space_violations[
            [
                "jpin",
                "final_suggestion",
                "max_allocated_space"
            ]
        ]
    )
```

```python
# ============================================================
# CELL 33 — VALIDATION: Negative suggestions
# ============================================================

negative_orders = df[
    df["final_suggestion"] < 0
]

print(
    "Negative order suggestions:",
    len(negative_orders)
)
```

```python
# ============================================================
# CELL 34 — VALIDATION: Units vs cases
# ============================================================

case_mismatch = df[
    df["final_suggestion"]
    != (
        df["final_cases_suggestion"]
        * df["case_size"].fillna(0)
    )
]

print(
    "Case/unit mismatches:",
    len(case_mismatch)
)

if len(case_mismatch) > 0:
    display(case_mismatch.head())
```

```python
# ============================================================
# CELL 35 — VALIDATION: Database
# ============================================================

db_count = pd.read_sql(
    "SELECT COUNT(*) AS count FROM replenishment_data",
    conn
).iloc[0]["count"]

print("CSV rows:", len(df))
print("SQLite rows:", db_count)

assert db_count == len(df), \
    "Database row count does not match CSV!"

print("Database row count validation PASSED.")
```

```python
# ============================================================
# CELL 36 — FINAL VALIDATION REPORT
# ============================================================

validation_report = pd.DataFrame({
    "Check": [
        "Input row count",
        "Database row count",
        "Negative suggestions",
        "Case rounding violations",
        "Space violations",
        "Case/unit mismatches",
        "MOV failures",
        "SKUs requiring replenishment"
    ],

    "Result": [
        len(df),
        db_count,
        len(negative_orders),
        len(case_violations),
        len(space_violations),
        len(case_mismatch),
        (df["mov_check"] == "FAIL").sum(),
        (df["final_suggestion"] > 0).sum()
    ]
})

display(validation_report)
```

```python
# ============================================================
# CELL 37 — BUSINESS SUMMARY
# ============================================================

total_skus = len(df)

replenishment_skus = (
    df["final_suggestion"] > 0
).sum()

total_units = df["final_suggestion"].sum()

total_cases = df["final_cases_suggestion"].sum()

total_value = df["final_value"].sum()

total_tonnage = df["final_tonnage"].sum()

mov_pass = (
    df["mov_check"] == "PASS"
).sum()

mov_fail = (
    df["mov_check"] == "FAIL"
).sum()

print("=" * 60)
print("REPLENISHMENT PLANNER SUMMARY")
print("=" * 60)

print(f"Total SKUs                  : {total_skus:,}")
print(f"SKUs requiring replenishment: {replenishment_skus:,}")
print(f"Total suggested units       : {total_units:,.0f}")
print(f"Total suggested cases       : {total_cases:,.0f}")
print(f"Total suggested value       : ₹{total_value:,.2f}")
print(f"Total suggested tonnage     : {total_tonnage:,.3f}")
print(f"MOV PASS                    : {mov_pass:,}")
print(f"MOV FAIL                    : {mov_fail:,}")

print("=" * 60)
```
